Task 2: Feature Extraction and Image Matching
Aim: To extract and match image features using HOG and SIFT algorithms.

Input: Two original photographs of the same person and scene taken with different framing.

Objective: To visualize HOG features, detect SIFT keypoints, match features between the two photographs, and compare results after rotation or resizing.

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from skimage.feature import hog

img1 = cv2.imread("input/photo1.jpeg")
img2 = cv2.imread("input/photo2.jpeg")

if img1 is None or img2 is None:
    raise FileNotFoundError("Check photo1.jpeg and photo2.jpeg")

gray1 = cv2.cvtColor(img1, cv2.COLOR_BGR2GRAY)
gray2 = cv2.cvtColor(img2, cv2.COLOR_BGR2GRAY)

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(img1, cv2.COLOR_BGR2RGB))
plt.title("Photo 1")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(img2, cv2.COLOR_BGR2RGB))
plt.title("Photo 2")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
# HOG Feature Extraction

small = cv2.resize(gray1, (256, 256))

features, hog_image = hog(
    small,
    orientations=9,
    pixels_per_cell=(8, 8),
    cells_per_block=(2, 2),
    visualize=True
)

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(small, cmap="gray")
plt.title("Original Grayscale")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(hog_image, cmap="gray")
plt.title("HOG Features")
plt.axis("off")

plt.tight_layout()
plt.show()

plt.imsave("output/hog_features.jpg", hog_image, cmap="gray")

print("Total HOG features:", len(features))
```

```python
small = cv2.resize(gray2, (256, 256))

features, hog_image = hog(
    small,
    orientations=9,
    pixels_per_cell=(8, 8),
    cells_per_block=(2, 2),
    visualize=True
)

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(small, cmap="gray")
plt.title("Original Grayscale - Photo 2")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(hog_image, cmap="gray")
plt.title("HOG Features")
plt.axis("off")

plt.show()

plt.imsave("output/hog_features.jpg", hog_image, cmap="gray")

print("Total HOG features:", len(features))
```

```python
# SIFT Feature Extraction

sift = cv2.SIFT_create()

kp1, des1 = sift.detectAndCompute(gray1, None)
kp2, des2 = sift.detectAndCompute(gray2, None)

# Draw keypoints on Photo 2
sift_image = cv2.drawKeypoints(
    img2, kp2, None,
    flags=cv2.DRAW_MATCHES_FLAGS_DRAW_RICH_KEYPOINTS
)

plt.figure(figsize=(8, 6))
plt.imshow(cv2.cvtColor(sift_image, cv2.COLOR_BGR2RGB))
plt.title("SIFT Keypoints - Photo 2")
plt.axis("off")
plt.show()

cv2.imwrite("output/sift_keypoints.jpg", sift_image)

print("Photo 1 Keypoints:", len(kp1))
print("Photo 2 Keypoints:", len(kp2))
print("Photo 2 Descriptor Shape:", des2.shape if des2 is not None else None)
```

```python
# SIFT Feature Matching

bf = cv2.BFMatcher()

matches = bf.knnMatch(des1, des2, k=2)

good_matches = []

for m, n in matches:
    if m.distance < 0.75 * n.distance:
        good_matches.append(m)

matched_image = cv2.drawMatches(
    img1, kp1,
    img2, kp2,
    good_matches, None,
    flags=cv2.DrawMatchesFlags_NOT_DRAW_SINGLE_POINTS
)

plt.figure(figsize=(15, 7))
plt.imshow(cv2.cvtColor(matched_image, cv2.COLOR_BGR2RGB))
plt.title("SIFT Matching - Photo 1 vs Photo 2")
plt.axis("off")
plt.show()

cv2.imwrite("output/sift_matching.jpg", matched_image)

print("Good Matches:", len(good_matches))
```

```python
# Rotate Photo 2 by 15 degrees

h, w = img2.shape[:2]

matrix = cv2.getRotationMatrix2D((w / 2, h / 2), 15, 1)

rotated = cv2.warpAffine(img2, matrix, (w, h))

gray_rotated = cv2.cvtColor(rotated, cv2.COLOR_BGR2GRAY)

kp_rot, des_rot = sift.detectAndCompute(gray_rotated, None)

matches_rot = bf.knnMatch(des1, des_rot, k=2)

good_rot = []

for m, n in matches_rot:
    if m.distance < 0.75 * n.distance:
        good_rot.append(m)

rotated_matching = cv2.drawMatches(
    img1, kp1,
    rotated, kp_rot,
    good_rot, None,
    flags=cv2.DrawMatchesFlags_NOT_DRAW_SINGLE_POINTS
)

plt.figure(figsize=(15, 7))
plt.imshow(cv2.cvtColor(rotated_matching, cv2.COLOR_BGR2RGB))
plt.title("SIFT Matching After Rotation")
plt.axis("off")
plt.show()

cv2.imwrite("output/sift_rotated_matching.jpg", rotated_matching)

print("Good Matches After Rotation:", len(good_rot))
```
