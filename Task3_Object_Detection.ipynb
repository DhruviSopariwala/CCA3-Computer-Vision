```python
import cv2
import matplotlib.pyplot as plt
from ultralytics import YOLO
import os

os.makedirs("output", exist_ok=True)

img = cv2.imread("input/photo2.jpeg")

if img is None:
    raise FileNotFoundError("Check input/photo2.jpeg")

plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Original Photo 2")
plt.axis("off")
plt.show()
```

```python
# Haar Cascade Face Detection

face_cascade = cv2.CascadeClassifier(
    cv2.data.haarcascades + "haarcascade_frontalface_default.xml"
)

gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

faces = face_cascade.detectMultiScale(
    gray,
    scaleFactor=1.1,
    minNeighbors=5
)

haar_img = img.copy()

for (x, y, w, h) in faces:
    cv2.rectangle(haar_img, (x, y), (x+w, y+h), (0, 255, 0), 2)
    cv2.putText(haar_img, "Face", (x, max(0, y-10)),
                cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0, 255, 0), 2)

plt.imshow(cv2.cvtColor(haar_img, cv2.COLOR_BGR2RGB))
plt.title("Viola-Jones Face Detection")
plt.axis("off")
plt.show()

cv2.imwrite("output/haar_detection.jpg", haar_img)

print("Faces detected:", len(faces))
```

```python
import cv2
import matplotlib.pyplot as plt
import os

os.makedirs("output", exist_ok=True)

img = cv2.imread("input/photo2.jpeg")

if img is None:
    raise FileNotFoundError("Check photo2.jpeg in inputs folder")

plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Original Photo 2")
plt.axis("off")
plt.show()
```

```python
# Haar Cascade Face Detection

face_cascade = cv2.CascadeClassifier(
    cv2.data.haarcascades + "haarcascade_frontalface_default.xml"
)

gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

faces = face_cascade.detectMultiScale(
    gray,
    scaleFactor=1.1,
    minNeighbors=5
)

haar_img = img.copy()

for (x, y, w, h) in faces:
    cv2.rectangle(haar_img, (x, y), (x+w, y+h), (0, 255, 0), 2)
    cv2.putText(
        haar_img, "Face", (x, max(0, y-10)),
        cv2.FONT_HERSHEY_SIMPLEX, 0.7,
        (0, 255, 0), 2
    )

plt.figure(figsize=(7, 7))
plt.imshow(cv2.cvtColor(haar_img, cv2.COLOR_BGR2RGB))
plt.title("Viola-Jones Face Detection")
plt.axis("off")
plt.show()

cv2.imwrite("output/haar_detection.jpg", haar_img)

print("Faces Detected:", len(faces))
```

```python
from ultralytics import YOLO

model = YOLO("yolo11n.pt")

results = model(img, verbose=False)

yolo_img = results[0].plot()

plt.figure(figsize=(7, 7))
plt.imshow(cv2.cvtColor(yolo_img, cv2.COLOR_BGR2RGB))
plt.title("YOLO Object Detection")
plt.axis("off")
plt.show()

cv2.imwrite("output/yolo_detection.jpg", yolo_img)

print("Objects Detected:", len(results[0].boxes))

for box in results[0].boxes:
    class_id = int(box.cls[0])
    confidence = float(box.conf[0])
    print(f"{model.names[class_id]}: {confidence:.2f}")
```

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(haar_img, cv2.COLOR_BGR2RGB))
plt.title("Viola-Jones")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(yolo_img, cv2.COLOR_BGR2RGB))
plt.title("YOLO")
plt.axis("off")

plt.tight_layout()
plt.savefig("output/detection_comparison.jpg")
plt.show()
```

### Task 3: Observation and Conclusion

**Aim:** To implement object recognition using Viola–Jones and YOLO.

**Input:** Original Photo 2 containing a person standing near a waterfall.

**Observation:**

The Viola–Jones algorithm was implemented using a pretrained Haar Cascade classifier. It successfully detected one face in the photograph and displayed a bounding box around it.

The YOLO algorithm was implemented using a pretrained object detection model. It successfully detected one person with a confidence score of **0.89 (89%)**.

**Comparison:**

- Viola–Jones detected the human face using Haar-like features and a cascade classifier.
- YOLO detected the complete person using a deep learning model.
- Both algorithms successfully identified their intended targets.
- YOLO provided a class label and confidence score, while the Haar Cascade implementation displayed a face label without a confidence score.

**Missed or Incorrect Detections:**

No incorrect detections were observed for the intended face and person targets in this experiment. The waterfall and rocks were not detected as objects by YOLO, but these are not necessarily supported object classes in the pretrained model.

**Conclusion:**

Both Viola–Jones and YOLO successfully performed object detection on the selected photograph. Viola–Jones was effective for detecting the face, whereas YOLO successfully detected the person with a confidence score of 89%. The experiment demonstrated the differences between traditional cascade-based detection and modern deep learning-based object detection.
